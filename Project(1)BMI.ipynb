```python
# For Jupyter Notebook (ensures plots show properly)
%matplotlib inline

import matplotlib.pyplot as plt

def calculate_bmi(weight, height, unit='imperial'):
    """Calculate BMI based on the given unit system."""
    if unit == 'metric':
        bmi = weight / (height ** 2)
    else:  # Default to imperial
        bmi = (weight * 703) / (height ** 2)
    return bmi

def categorize_bmi(bmi):
    """Categorize BMI into weight classes."""
    if bmi < 18.5:
        return 'Underweight'
    elif bmi < 25:
        return 'Normal weight'
    elif bmi < 30:
        return 'Overweight'
    elif bmi < 35:
        return 'Obese'
    elif bmi < 40:
        return 'Severely Obese'
    else:
        return 'Morbidly Obese'

def main(name="Test User", unit="imperial", weight=150, height=65):
    """Main function to calculate BMI and visualize it."""
    
    # Input validation
    if weight <= 0 or height <= 0:
        print("Error: Weight and height must be positive numbers.")
        return
    
    # Calculate BMI
    bmi = calculate_bmi(weight, height, unit)
    category = categorize_bmi(bmi)
    
    # Print result
    print(f"{name}, your BMI is {bmi:.2f} ({category}).")
    
    # Visualization
    categories = ['Underweight', 'Normal weight', 'Overweight', 
                  'Obese', 'Severely Obese', 'Morbidly Obese']
    bmi_values = [18.5, 25, 30, 35, 40, 45]
    
    plt.figure(figsize=(8, 5))
    plt.bar(categories, bmi_values)
    plt.axhline(y=bmi, linestyle='--', label=f'Your BMI: {bmi:.2f}')
    
    plt.xlabel("BMI Categories")
    plt.ylabel("BMI Value")
    plt.title("BMI Classification")
    plt.xticks(rotation=30)
    plt.legend()
    
    plt.show()


main(name="Tisha", unit="metric", weight=58, height=1.57)
```
